### Handling tipe data

```python
superstore_copy["Order Date"] = pd.to_datetime(superstore_copy["Order Date"])
superstore_copy["Ship Date"] = pd.to_datetime(superstore_copy["Ship Date"])
superstore_copy["Postal Code"]=superstore_copy["Postal Code"].astype(str)
superstore_copy.info()
```

### Menangani outlier

```python
kolom_outlier = superstore_copy.select_dtypes(include=["float64"]).columns
```

```python
kolom_outlier
```

```python
for i in kolom_outlier :
    Q1 = superstore_copy[i].quantile(0.25)
    Q3 = superstore_copy[i].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outlier = (superstore_copy[i] < lower) | (superstore_copy[i] > upper)
    jumlah_outlier = outlier.sum()
    persentase_outlier = jumlah_outlier/len(outlier) * 100
    print(f"Jumlah Outlier pada Kolom {i} : {jumlah_outlier}")
    print(f"Persentase Outlier pada Kolom {i} : {persentase_outlier}\n")
```
